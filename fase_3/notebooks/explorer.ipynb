# 📚 Dicionário de Dados — Medical Abstracts TC Corpus

Este dataset contém resumos médicos (*abstracts*) em língua inglesa divididos em tarefas supervisionadas de Classificação Multiclasse de Texto (NLP). O conjunto é composto por 3 arquivos CSV:

1. `medical_tc_train.csv` (Conjunto de Treino)
2. `medical_tc_test.csv` (Conjunto de Teste)
3. `medical_tc_labels.csv` (Tabela de De-Para de Classes)

---

## 1. DataFrames de Dados (`medical_tc_train.csv` e `medical_tc_test.csv`)

Contêm as instâncias de treino (~11.550 registros) e teste (~2.888 registros).

| Nome da Coluna | Tipo de Dado | Descrição | Exemplo de Valor |
| :--- | :--- | :--- | :--- |
| `condition_label` | `int64` / Categórico | Identificador numérico inteiro (1 a 5) que representa a classe/condição médica à qual o resumo pertence. Atua como a **variável alvo (*target*)**. | `1`, `4` |
| `medical_abstract` | `string` / Texto | Texto completo e não formatado do resumo médico (*abstract*). Atua como a **variável preditora (*feature*)**. | `"The effect of antihypertensive therapy on..."` |

---

## 2. DataFrame de Metadados (`medical_tc_labels.csv`)

Mapeia o código numérico presente na coluna `condition_label` para a sua respectiva condição clínica.

| Nome da Coluna | Tipo de Dado | Descrição | Exemplo de Valor |
| :--- | :--- | :--- | :--- |
| `condition_label` | `int64` / Chave Primária | Código numérico identificador da classe (de 1 a 5). | `1`, `2` |
| `condition_name` | `string` | Nome descritivo da condição médica ou sistema patológico. | `"neoplasms"`, `"cardiovascular diseases"` |

---

## 3. Mapeamento e Descrição das Classes

| `condition_label` | `condition_name` | Descrição Clínica |
| :---: | :--- | :--- |
| **1** | `neoplasms` | Resumos relacionados a tumores, cânceres, oncologia, biópsias, radioterapia e quimioterapia. |
| **2** | `digestive system diseases` | Doenças do trato gastrointestinal, fígado, pâncreas, vesícula biliar e intestino. |
| **3** | `nervous system diseases` | Condições neurológicas (cérebro, medula espinhal, nervos periféricos, AVC, demências, etc.). |
| **4** | `cardiovascular diseases` | Doenças cardíacas e vasculares (infarto, hipertensão, arritmias, insuficiência cardíaca). |
| **5** | `general pathological conditions` | Condições sistêmicas, processos inflamatórios/infecciosos gerais ou sintomas inespecíficos. |

---

## Importação de blibiotecas

````python
import os
import kagglehub
import pandas as pd
from sklearn.svm import LinearSVC
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import classification_report, accuracy_score
````

## Ingestão de dados

````python

# Carrega os dataframes de treino, teste e labels do dataset Medical Abstracts TC Corpus.
    
path = kagglehub.dataset_download("saharalaa/medical-abstracts-tc-corpus")

print("Path to dataset files:", path)

print(os.listdir(path))

# Load the train, test, and labels dataframes
df_train = pd.read_csv(f"{path}/medical_tc_train.csv")
df_test = pd.read_csv(f"{path}/medical_tc_test.csv")
df_labels = pd.read_csv(f"{path}/medical_tc_labels.csv")   

print("Path to dataset files:", path)
````

````python
label_mapping = dict(zip(df_labels['condition_label'], df_labels['condition_name']))
print("Label mapping:", label_mapping)
````

````python
X_train = df_train['medical_abstract']
y_train = df_train['condition_label']

X_test = df_test['medical_abstract']
y_test = df_test['condition_label']
````

````python
# Adicionamos stopwords clínicas genéricas para reduzir ruído
clinical_stopwords = list(TfidfVectorizer(stop_words='english').get_stop_words()) + [
    'patient', 'patients', 'study', 'results', 'case', 'cases', 'group', 'method', 'methods'
]
````

````python
# Configuração otimizada do TF-IDF para máxima acurácia e leveza
tfidf_params = {
    "stop_words": clinical_stopwords,
    "lowercase": True,
    "strip_accents": "unicode",
    "ngram_range": (1, 1),
    "min_df": 5,
    "max_df": 0.95,
    "sublinear_tf": False
}
````

````python
# Pipeline otimizada com LinearSVC (Support Vector Classifier)
# Hiperparâmetro C calibrado (C=0.06) para generalização máxima em alta dimensionalidade
pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(**tfidf_params)),
    ('clf', LinearSVC(
        C=0.06,
        random_state=42,
        dual=False
    ))
])
````

````python
print("Predizendo rótulos para o conjunto de teste...")
y_pred = pipeline.predict(X_test)
````

````python
print("predict shape:", y_pred.shape)
````

````python
print("=== RELATÓRIO DE AVALIAÇÃO (MODELO OTIMIZADO) ===")
acc = accuracy_score(y_test, y_pred)
print(f"Acurácia Global: {acc:.4f} ({acc*100:.2f}%)\n")
print(classification_report(y_test, y_pred, target_names=df_labels['condition_name'].tolist()))
````
