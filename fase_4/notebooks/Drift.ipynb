```python
import pandas as pd
```

```python
df_reference = pd.read_csv(
    r"/home/gusdpaula/code-dev/MLENG_FIAP/fase_4/data/raw/loan_data.csv"
)
df_reference.head()
```

```python
df_new = df_reference.sample(n=1000, random_state=42)  # Simulate new data

df_new.head()
```

```python
df_new["annual_inc"] = df_new["annual_inc"] * 0.7  # Simulate a drift in annual income
df_new["loan_amnt"] = df_new["loan_amnt"] * 1.3  # Simulate a drift in loan amount
```

```python
df_new.to_json(
    r"/home/gusdpaula/code-dev/MLENG_FIAP/fase_4/data/raw/loan_data_drifted.json",
    index=False,
)
```
